```python
# pandas is used for data handling and cleaning
# numpy is used for numerical operations
import pandas as pd
import numpy as np
```

```python
# Load the uploaded CSV file into a DataFrame
df = pd.read_csv("/customers-100.csv")

# Display first 5 rows to understand data
df.head()
```

```python
# Clean column names (BEST PRACTICE)
df.columns = df.columns.str.strip().str.lower()

# See actual column names
df.columns
```

```python
# Check column names, data types, and missing values
df.info()
```

```python
# Check number of rows and columns
df.shape
```

```python
# Count duplicate rows
df.duplicated().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Confirm duplicates are removed
df.duplicated().sum()
```

```python
# Remove unnecessary columns like temp IDs or notes
# errors='ignore' avoids error if column does not exist
df = df.drop(columns=["temp_id", "notes"], errors="ignore")
```

```python
# Rename columns for clarity
df = df.rename(columns={
    "cust_nm": "customer_name",
    "reg": "region"
})
```

```python
# Check missing values column-wise
df.isna().sum()
```

```python
# Drop rows where mandatory field ('customer id') is missing
df = df.dropna(subset=["customer id"])
```

```python
# Convert date column from string to datetime format
df["subscription date"] = pd.to_datetime(df["subscription date"], errors="coerce")
```

```python
# Create a 'customer_name' column by combining 'first name' and 'last name'
df['customer_name'] = df['first name'] + ' ' + df['last name']
# Convert names to lowercase and remove extra spaces
df["customer_name"] = df["customer_name"].str.lower().str.strip()
```

```python
# Normalize text columns: lowercase and strip whitespace
text_cols = ['company', 'city', 'country', 'email', 'website', 'first name', 'last name']

for col in text_cols:
    if col in df.columns and df[col].dtype == 'object': # Check if column exists and is of object type
        df[col] = df[col].str.lower().str.strip()

# Display a sample of the normalized data to verify
print("Text columns normalized. Here's a look at some of them:")
display(df[['company', 'city', 'country', 'email']].head())
```

```python
# Ensure no missing values remain
assert df.isna().sum().sum() == 0
```

```python
# Ensure no duplicate records remain
assert df.duplicated().sum() == 0
```

```python
# Save cleaned dataset to a new CSV file
df.to_csv("cleaned_dataset.csv", index=False)
```
